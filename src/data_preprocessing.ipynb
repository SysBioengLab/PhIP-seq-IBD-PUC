```python
import pandas as pd
```

```python
def mapper(x):
    if x == 'Never':
        return 0
    elif x == 'Current_smoker':
        return 1
    else :
        return 2
metadata2 = pd.read_csv('../Data/metadata2.csv')
metadata2['type_person'] = metadata2['type_person'].apply(lambda x: 0 if x == 'Non_IBD' else 1)
metadata2['smoking'] = metadata2['smoking'].apply(lambda x: mapper(x))
metadata2['Sex'] = metadata2['Sex'].apply(lambda x: 1 if x == 'Female' else 0)
metadata2['appendectomy'] = metadata2['appendectomy'].apply(lambda x: 1 if x == 'Yes' else 0)

# tes
test = [107,  86,  48,  43,  74, 108,  97,  90,  17,   7]
metadata2_test = metadata2.loc[test]
metadata2_test.index = range(len(metadata2_test))
metadata2_test.to_csv('../Data/metadata2_test.csv', index=False)
metadata2_train = metadata2.drop(test)
metadata2_train.index = range(len(metadata2_train))
metadata2_train.to_csv('../Data/metadata2_train.csv', index=False)
```
